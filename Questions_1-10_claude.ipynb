##### **Performing EDA And Solving Some Questions Using Python , Pandas And Numpy**

```python
# Import required libraries
# re      -> for regex-based text extraction (used in the feature-extraction stage)
# pandas  -> for loading, filtering, and analyzing tabular data
# numpy   -> for numerical/statistical operations
import re
import pandas as pd
import numpy as np
```

**Read csv File**

```python
# Load the cleaned dataset (produced in the feature-extraction stage)
# into a Pandas DataFrame called 'data'
data = pd.read_csv(r'updated_anime_dataset.csv')
```

**Top 5 Highest Rated Anime**

```python
# Q1: Top 5 Highest Rated Anime
# nlargest(5, 'Score') sorts the DataFrame by the 'Score' column
# and returns the 5 rows with the highest values
data.nlargest(5, 'Score')
```

**Top 5 Most Popular Anime**

```python
# Q2: Top 5 Most Popular Anime (by Members)
# Sort the entire DataFrame by 'Members' in descending order,
# then take the first 5 rows (head() defaults to 5)
data.sort_values('Members', ascending=False).head()
```

**Lowest Member Count**

```python
# Q3: Anime With the Lowest Member Count
# Find the minimum value in the 'Members' column, then filter
# the DataFrame to the row(s) matching that minimum,
# keeping only the 'Title' and 'Members' columns
data[data['Members'] == data['Members'].min()][['Title', 'Members']]
```

**Score Metrices**

```python
# Q4: Score Metrics (Method 1 - using NumPy)
# Either Use The Following 5
# Compute mean, median, standard deviation, min, and max of 'Score'
# individually using NumPy functions

print(np.mean(data['Score']))    # average score
print(np.median(data['Score']))  # middle value when sorted
print(np.std(data['Score']))     # spread/variability of scores
print(data['Score'].min())       # lowest score
print(data['Score'].max())       # highest score
```

```python
# Q4: Score Metrics (Method 2 - shortcut)
# Or Just Use This One
# describe() gives count, mean, std, min, quartiles, and max
# in a single summary — same info as above, computed automatically

data['Score'].describe()
```

**High-Score & High-Popularity**

```python
# Q5: High-Score & High-Popularity Anime
# Return Anime Title, Score, and Members where score >= 9.00
# and Members > 1 Million
# The '&' combines both boolean conditions (row must satisfy both)

data[(data['Score'] >= 9.00) & (data['Members'] > 1000000)][['Title', 'Score', 'Members']]
```

**Count Long-Running Anime**

```python
# Q6: Count Long-Running Anime
# Animes Where Episodes > 100
# Filter rows where 'Episodes' exceeds 100, then count
# how many titles satisfy that condition

print(data[data['Episodes'] > 100]['Title'].count())
```

**What Are Long-Running Anime**

```python
# Q7: What Are the Long-Running Anime
# Same filter as Q6, but this time display the full rows
# instead of just a count
data[data['Episodes'] > 100]
```

**Short-Series High Performers**

```python
# Q8: Short-Series High Performers
# anime ki list nikalein jinke episodes <= 13 hain lekin score >= 9.00 hai
# (List anime with 13 or fewer episodes but a score of 9.00 or higher)
# Combines two conditions with '&' to find short but highly-rated anime

data[(data['Episodes'] <= 13) & (data['Score'] >= 9.00)]
```

**Episode Count Using Custom Grouping**

```python
# Q9: Episode Count Using Custom Grouping (setup)
# Define bin edges and matching labels to group anime by episode count:
#   0-1        -> '1 Episode'
#   1-13       -> '2-13 Episodes'
#   13-26      -> '14-26 Episodes'
#   26-infinity-> '27+ Episodes'
bins = [0, 1, 13, 26, float('inf')]

labels = [
    '1 Episode',
    '2-13 Episodes',
    '14-26 Episodes',
    '27+ Episodes'
]
```

```python
# Create a new 'Episode_Bin' column by cutting the 'Episodes' column
# into the ranges defined above using pd.cut()
# include_lowest=True ensures the very first value (0) is included in a bin
data['Episode_Bin'] = pd.cut(
    data['Episodes'],
    bins=bins,
    labels=labels,
    include_lowest=True
)
```

```python
# Count how many anime fall into each Episode_Bin category,
# sorted from smallest to largest count
data['Episode_Bin'].value_counts().sort_values()
```

**Average Score by Episode Bin**

```python
# Q10: Average Score by Episode Bin
# Group the data by the 'Episode_Bin' category, compute the mean
# 'Score' within each group, and sort groups from highest to lowest average
data.groupby('Episode_Bin')['Score'].mean().sort_values(ascending=False)
```
