#### **Import Required Libraries**

```python
import re
import numpy as np
import pandas as pd
```

#### **Load Raw Dataset**

```python
data = pd.read_csv(r'anime.csv')
```

#### **Extract Episodes Count From Title**

```python
# Loop Through Each Character Of Title And Capture Text Between Parenthesis, Since Episode Count Is Always Written Inside Parenthesis

def extract_episodes(txt):
    check = False
    ep_data = ""

    for i in txt:
        if i == ")":
            return ep_data
        if check == True:
            ep_data += i
        if i == "(":
            check = True
```

```python
# Apply Function To Get Raw Episode Text, Then Remove The Word " eps" And Convert Result Into Integer Type

data['Episodes'] = data['Title'].apply(extract_episodes).str.replace(" eps","").astype(int)
data['Episodes']
```

#### **Extract Timespan From Title**

```python
# Split Title On Closing Parenthesis And Take The Part After It, Since Timespan Always Starts After Episode Info And Is Exactly 19 Characters Long

def extract_timespan(txt):
    if ')' in txt:
        after_bracket = txt.split(')')[1]
        return after_bracket[:19].strip()
    return None
```

```python
# Approach: Apply Function On Title Column To Create New Timespan Column

data['Timespan'] = data['Title'].apply(extract_timespan)
data['Timespan']
```

#### **Extract Members From Title**

```python
# Approach: Skip Past The Episode Info And Timespan Part Of The String , Then Use Regex To Find The Numeric Value Followed By The Word "members"

def extract_members(txt):
    if ')' in txt:
        after_bracket = txt.split(')')[1]
        after_timespan = after_bracket[19:]  
        match = re.search(r'([\d,]+)\s*members', after_timespan)
        if match:
            return int(match.group(1).replace(',', ''))
    return None
```

```python
# Approach: Apply Function To Extract Raw Match, Then Remove Commas And Convert Result Into Integer Type

data['Members'] = data['Title'].apply(extract_members).astype(int)  
data['Members']
```

#### **Save Updated Dataset**

```python
# Approach: Export Final Dataframe Containing Original Columns Plus Newly Extracted Features Into A New CSV File

data.to_csv('updated_anime_dataset.csv', index=False)
```
