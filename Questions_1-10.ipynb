##### **Performing EDA And Solving Some Questions Using Python , Pandas And Numpy**

```python
# Import Required Libraries

import pandas as pd
import numpy as np
```

**Read csv File**

```python
# Load The Cleaned Dataset 

data = pd.read_csv(r'updated_anime_dataset.csv')
```

**Top 5 Highest Rated Anime**

```python
# nlargest(5, 'Score') Sorts The DataFrame By The 'Score' Column And Returns The 5 Rows With The Highest Values

data.nlargest(5, 'Score')
```

**Top 5 Most Popular Anime**

```python
# Sort DataFrame By 'Members' In Descending Order , Then Take The First 5 Rows , head() Gives 5 Rows By Default

data.sort_values('Members',ascending=False).head()
```

**Lowest Member Count**

```python
# Find The Minimum Value In The 'Members' Column , Then Filter The DataFrame To The Row Matching That Minimum
# Keeping Only The 'Title' And 'Members' Columns

data[data['Members'] == data['Members'].min()][['Title','Members']]
```

**Score Metrices**

```python
# Compute Mean, Median, Standard Deviation, Min, And Max Of 'Score'
# Either Use The Following 5 

print(np.mean(data['Score']))
print(np.median(data['Score']))
print(np.std(data['Score']))
print(data['Score'].min())
print(data['Score'].max())
```

```python
# Or Just Use This One 
# Describe() Gives Count, Mean, Std, Min, Quartiles, And Max

data['Score'].describe()
```

**High-Score & High-Popularity**

```python
# Return Anime Title, Score, And Members Where Score >= 9.00 And Members > 1 Million
# The '&' Combines Both Boolean Conditions (Row Must Satisfy Both)

data[(data['Score'] >= 9.00) & (data['Members'] > 1000000) ][['Title','Score','Members']]
```

**Count Long-Running Anime**

```python
# Animes Where Episodes > 100 Filter Rows Where 'Episodes' Exceeds 100
# Then Count How Many Titles Satisfy That Condition

print(data[data['Episodes'] > 100]['Title'].count())
```

**What Are Long-Running Anime**

```python
# Same Filter As Q6
# But This Time Display The Full Rows Instead Of Just A Count

data[data['Episodes'] > 100]
```

**Short-Series High Performers**

```python
# List Anime With 13 Or Fewer Episodes But A Score Of 9.00 Or Higher
# Combines Two Conditions With '&' To Find Short But Highly-Rated Anime

data[ (data['Episodes'] <= 13) & (data['Score'] >= 9.00) ]
```

**Episode Count Using Custom Grouping**

```python
# Define Bin Edges And Matching Labels To Group Anime By Episode Count

bins = [0, 1, 13, 26, float('inf')]

labels = ['1 Episode', '2-13 Episodes', '14-26 Episodes', '27+ Episodes']
```

```python
# Create A New 'Episode_Bin' Column By Cutting The 'Episodes' Column Into The Ranges Defined Above Using Pd.cut()
# Include_lowest=True Ensures The Very First Value (0) Is Included In A Bin

data['Episode_Bin'] = pd.cut(
    data['Episodes'],
    bins=bins,
    labels=labels,
    include_lowest=True
)
```

```python
# Count How Many Anime Fall Into Each Episode_Bin Category, Sorted From Smallest To Largest Count

data['Episode_Bin'].value_counts().sort_values()
```

**Average Score by Episode Bin**

```python
# Group The Data By The 'Episode_Bin' Category, Compute The Mean 'Score' Within Each Group, And Sort Groups From Highest To Lowest Average

data.groupby('Episode_Bin')['Score'].mean().sort_values(ascending=False)
```
